```python
#W8a3
def is_happy(n):
    seen = set()
    while n != 1:
        if n in seen:
            return False
        seen.add(n)
        n = sum(int(digit) ** 2 for digit in str(n))
    return True

try:
    a_input = input("Nhập vào một số nguyên dương a: ")
    a = int(a_input)
    
    if a <= 0:
        print("NO")
    else:
        if is_happy(a):
            print("YES")
        else:
            print("NO")
            
except ValueError:
    print("Lỗi: Đầu vào không phải là một số nguyên hợp lệ!")
```

```python
#w8a4
try:
    a,b = map(float,input().split())
except ValueError:
    print("Lỗi: Bạn đã nhập sai kiểu dữ liệu! Vui lòng nhập số.")
else:
    tong = a + b
    print(f"Tổng hai số là: {tong}")
finally:
    print("Kết thúc chương trình.")
```

```python
#w8a5
danh_sach = list(map(str,input().split()))
ket_qua = []

for x in danh_sach:
    try:
        ket_qua.append(int(x))
    except ValueError:
        print(f"Cảnh báo: Không thể chuyển '{x}' sang int. Bỏ qua.")

print(ket_qua)
```
