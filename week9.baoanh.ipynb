```python
#w9a1
def mergeSort(arr1, arr2):
    res = []
    while arr1 and arr2:
        if arr1[0] < arr2[0]:
            res.append(arr1.pop(0))
        else:
            res.append(arr2.pop(0))
    res.extend(arr1)
    res.extend(arr2)
    return res
```

```python
#w9a2
def nthSmallest(mat, n):
    m = len(mat) - 1
    n = len(mat[0]) - 1
    min = mat[0][0]
    max = mat[-1][-1]
    ans = None
    while min <= max:
        mid = (min + max) // 2
        i = 0
        j = n
        count = 0
        while i <= m:
            if mat[i][j] > mid:
                j -= 1
                if j < 0:
                    break
                else:
                    count += j + 1
                    i += 1
                    if i > m:
                        break
        if count >= n:
            ans = mid
            max = mid - 1
        else:
            min = mid + 1
    return ans
```

```python
#w9a3
def swap2Digits(number):
    number = str(number)
    result = list(number)
    digits = sorted(list(number))
    for i in range(len(digits)):
        if number[i] != digits[i]:
            result[i], result[number.rindex(digits[i])] = result[number.rindex(digits[i])], result[i]
            break
    else:
        result[-1], result[-2] = result[-2], result[-1]
    return int("".join(result))
swap2Digits(1283920182731902374)
```

```python
#w9a4
def checkTransformable(a, b, depth=0):
    if len([c for c in a if c.isupper()]) >= len(b) and depth == 0:
        return False
    else:
        for i in range(len(a)):
            if a[i].upper() == b[0]:
                if len(b) == 1:
                    return True
                else:
                    return checkTransformable(a[i+1:], b[1:], depth+1)
        return False
```

```python
#w9a5
def checkSublist(list1, list2):
    for i in range(len(list2) - len(list1)):
        if list2[i:i+len(list1)] == list1:
            return True
    return False
```

```python
#w9a6
def nextPermutation(arr):
    for i in range(len(arr) - 1, 0, -1):
        if arr[i-1] < arr[i]:
            m = i - 1
            break
    else:
        return arr[::-1]
    for i in range(len(arr) - 1, m, -1):
        if arr[i] > arr[m]:
            arr[i], arr[m] = arr[m], arr[i]
            arr[m+1:] = arr[-1:m:-1]
            return arr
nextPermutation([1,4,2,6,7,4])
```

```python
#w9a8
def is_path_crossing(moves):
    x = y = 0
    visited = {(0, 0)}
    
    for move in moves:
        if move == 'U': y += 1
        elif move == 'D': y -= 1
        elif move == 'R': x += 1
        elif move == 'L': x -= 1
        
        if (x, y) in visited:
            return True
        visited.add((x, y))
        
    return False
print(is_path_crossing("RU"))       
print(is_path_crossing("UD"))      
print(is_path_crossing("RDRU"))    
print(is_path_crossing("RUULL"))   
print(is_path_crossing("RURDL"))   
```
