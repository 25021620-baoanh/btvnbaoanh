```python
#w8a1
import math
class cylinder:
    def __init__(self,r,h):
        self.bk = r
        self.cc = h
        self.pi = 3.14
    def surface_area(self):
        S= 2*self.bk*self.pi*(self.bk + self.cc)
        return S
    def volume(self):
        V= self.bk **2 * self.pi *self.cc
        return V
r,h=map(float, input().split())
c=cylinder(r,h)
print(f"{c.surface_area():.2f}  {c.volume():.2f}")  
```

```python
#w8a6
class Rectangle:
    def __init__(self, w, h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h
    def perimeter(self):
        return 2 * (self.w + self.h)
    def scale(self, k):
        self.w *= k
        self.h *= k
w,h=map(float,input().split())
a=Rectangle(w,h)
a.scale(6)
print(a.area())
print(a.perimeter())

```

```python
#w8a7
import math

class Fraction:
    def __init__(self, num, den):
        common = math.gcd(num, den)
        self.num = num // common
        self.den = den // common
        if self.den < 0:
            self.num = -self.num
            self.den = -self.den

    def add(self, other):
        return Fraction(self.num * other.den + other.num * self.den, self.den * other.den)

    def sub(self, other):
        return Fraction(self.num * other.den - other.num * self.den, self.den * other.den)

    def mul(self, other):
        return Fraction(self.num * other.num, self.den * other.den)

    def div(self, other):
        return Fraction(self.num * other.den, self.den * other.num)

    def __str__(self):
        return f"{self.num}/{self.den}"

try:
    data = input().split()
    a, b, op, c, d = int(data[0]), int(data[1]), data[2], int(data[3]), int(data[4])
    
    f1 = Fraction(a, b)
    f2 = Fraction(c, d)
    
    if op == '+': print(f1.add(f2))
    elif op == '-': print(f1.sub(f2))
    elif op == '*': print(f1.mul(f2))
    elif op == '/': print(f1.div(f2))
except:
    pass
```

```python
#w8a9
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = []

    def add_score(self, subject, score):
        self.scores.append(score)

    def avg(self):
        if not self.scores:
            return 0.0
        return sum(self.scores) / len(self.scores)

    def rank(self):
        average = self.avg()
        if average >= 8:
            return "Excellent"
        elif 6.5 <= average < 8:
            return "Good"
        elif 5 <= average < 6.5:
            return "Average"
        else:
            return "Poor"

    def display(self):
        print(f"{self.name} {self.avg():.2f} {self.rank()}")

def main():
    while True:
        try:
            line = input().strip()
            if not line:
                break
            
            name = line
            n = int(input().strip())
            
            student = Student(name)
            for _ in range(n):
                sub_data = input().split()
                # Lấy phần tử cuối làm điểm số, các phần tử trước là tên môn học
                score = float(sub_data[-1])
                subject = " ".join(sub_data[:-1])
                student.add_score(subject, score)
            
            student.display()
        except EOFError:
            break
        except Exception:
            break

if __name__ == "__main__":
    main()
```

```python
#W8a11
import math

class Point:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z

    def __sub__(self, other):
        return Point(self.x - other.x, self.y - other.y, self.z - other.z)

def cross_product(p1, p2):
    return Point(
        p1.y * p2.z - p1.z * p2.y,
        p1.z * p2.x - p1.x * p2.z,
        p1.x * p2.y - p1.y * p2.x
    )

def dot_product(p1, p2):
    return p1.x * p2.x + p1.y * p2.y + p1.z * p2.z

def magnitude(p):
    return math.sqrt(p.x**2 + p.y**2 + p.z**2)

def main():
    points = []
    for _ in range(4):
        coords = list(map(float, input().split()))
        points.append(Point(coords[0], coords[1], coords[2]))

    A, B, C, D = points

    # Vector pháp tuyến của mặt phẳng (A, B, C)
    AB = B - A
    BC = C - B
    n1 = cross_product(AB, BC)

    # Vector pháp tuyến của mặt phẳng (B, C, D)
    CD = D - C
    n2 = cross_product(BC, CD)

    # Tính góc PHI giữa hai vector pháp tuyến
    cos_phi = dot_product(n1, n2) / (magnitude(n1) * magnitude(n2))
    
    # Đảm bảo giá trị nằm trong khoảng [-1, 1] để tránh lỗi math domain
    cos_phi = max(-1.0, min(1.0, cos_phi))
    
    phi_rad = math.acos(cos_phi)
    phi_deg = math.degrees(phi_rad)

    print(f"{phi_deg:.2f}")

if __name__ == "__main__":
    main()
```
